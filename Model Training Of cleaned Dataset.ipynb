```python
#### SEperate independent and Dependent features 
x = df.drop('FWI',axis=1)
y = df['FWI']
```

```python
#### Train test split 
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.25,random_state=42)
```

```python
## Feature selection based on correlation  
df.corr()
```

```python
plt.figure(figsize=(15,10))
sns.heatmap(df.corr(),annot=True)
```

```python
#### check for multicollinearity  on trainig data 
plt.figure(figsize=(12,10))
corr = x_train.corr()
sns.heatmap(corr,annot=True)
```

```python
### remove multicollinearity features with threshold 
def correlation(dataset, threshold):
    col_corr = set()
    corr_matrix = dataset.corr()
    for i in range(len(corr_matrix.columns)):
        for j in range(i): 
            if abs(corr_matrix.iloc[i,j]) > threshold: 
                colname = corr_matrix.columns[i]
                col_corr.add(colname) 
    return col_corr           
```

```python
## threshold--Domain expertise
corr_features=correlation(x_train,0.85)
```

```python
corr_features
```

```python
## drop features when correlation is more than 0.85 
x_train.drop(corr_features,axis=1,inplace=True)
x_test.drop(corr_features,axis=1,inplace=True)
x_train.shape,x_test.shape
```

## Feature scaling or standardization

```python
from sklearn.preprocessing import StandardScaler 
scaler = StandardScaler() 
x_train_scaled = scaler.fit_transform(x_train) 
x_test_scaled = scaler.transform(x_test)
```

```python
x_train_scaled
```

## Box plot to understand the effect of sStandardScaler 

```python
plt.subplots(figsize=(15,5))
plt.subplot(1,2,1)
sns.boxplot(data=x_train)
plt.title('X_train Before Scaling')
plt.subplot(1,2,2)
sns.boxenplot(data=x_train_scaled)
plt.title('X_train After Scaling')
plt.show()
```

```python
### Linear Regression model 
from sklearn.linear_model import LinearRegression 
model = LinearRegression()
model.fit(x_train_scaled,y_train) 
y_pred = model.predict(x_test_scaled)
```

```python
y_pred
```

```python
### performance metrics 
from sklearn.metrics import r2_score, mean_absolute_error 
score = r2_score(y_test,y_pred)
mae = mean_absolute_error(y_test,y_pred) 
print(mae)
print(score)
plt.scatter(y_test,y_pred)
```

### Ridge Regression 


```python
from sklearn.linear_model import Ridge 
reduce_overfitting = Ridge()
reduce_overfitting.fit(x_train_scaled,y_train) 
y_pred_ridge =reduce_overfitting.predict(x_test_scaled)
```

```python
y_pred_ridge
```

```python
### performance metrics 
from sklearn.metrics import r2_score, mean_absolute_error 
score = r2_score(y_test,y_pred_ridge)
mae = mean_absolute_error(y_test,y_pred_ridge) 
print(mae)
print(score)
plt.scatter(y_test,y_pred_ridge)
```

#### Lasso Regression  - use in the case of multicollinearity

```python
from sklearn.linear_model import Lasso 
feature_selection = Lasso()
feature_selection.fit(x_train_scaled,y_train) 
y_pred_lasso =feature_selection.predict(x_test_scaled)
```

```python
y_pred_lasso
```

#### Cross Validation 

```python
### ridge 
from sklearn.linear_model import RidgeCV
ridge_cv = RidgeCV(alphas=[0.01, 0.1, 1, 10, 100])
ridge_cv.fit(x_train_scaled, y_train)
print(ridge_cv.alpha_)
```

```python
ridge_cv.alphas
```

```python
## lasso
from sklearn.linear_model import LassoCV
lasso_cv = LassoCV(alphas=[0.001, 0.01, 0.1, 1, 10], cv=5)
lasso_cv.fit(x_train_scaled, y_train)
print(lasso_cv.alpha_)
```

```python
## elasticnet
from sklearn.linear_model import ElasticNetCV
elastic_cv = ElasticNetCV(
    alphas=[0.001, 0.01, 0.1, 1, 10],
    l1_ratio=[0.1, 0.5, 0.9],
    cv=5
)
elastic_cv.fit(x_train_scaled, y_train)

print(elastic_cv.alpha_)
print(elastic_cv.l1_ratio_)
```
